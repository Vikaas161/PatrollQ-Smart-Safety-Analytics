```python
severity_map = {
    'HOMICIDE': 4,
    'CRIM SEXUAL ASSAULT': 4,
    'CRIMINAL SEXUAL ASSAULT': 4,
    'HUMAN TRAFFICKING': 4,
    'KIDNAPPING': 4,

    'ASSAULT': 3,
    'BATTERY': 3,
    'ROBBERY': 3,
    'WEAPONS VIOLATION': 3,
    'OFFENSE INVOLVING CHILDREN': 3,
    'STALKING': 3,
    'INTIMIDATION': 3,
    'ARSON': 3,
    'SEX OFFENSE': 3,

    'BURGLARY': 2,
    'MOTOR VEHICLE THEFT': 2,
    'CRIMINAL DAMAGE': 2,
    'THEFT': 2,
    'DECEPTIVE PRACTICE': 2,
    'CRIMINAL TRESPASS': 2,
    'OTHER OFFENSE': 2,
    'NARCOTICS': 2,
    'OTHER NARCOTIC VIOLATION': 2,
    'INTERFERENCE WITH PUBLIC OFFICER': 2,
    'CONCEALED CARRY LICENSE VIOLATION': 2,

    'PROSTITUTION': 1,
    'PUBLIC PEACE VIOLATION': 1,
    'LIQUOR LAW VIOLATION': 1,
    'GAMBLING': 1,
    'PUBLIC INDECENCY': 1,
    'OBSCENITY': 1,
    'NON-CRIMINAL': 1,
    'RITUALISM': 1
}

df['Crime_Severity_Score'] = df['Primary Type'].map(severity_map)
```

```python
def season(month):
    if month in [12,1,2]:
        return "Winter"
    elif month in [3,4,5]:
        return "Spring"
    elif month in [6,7,8]:
        return "Summer"
    else:
        return "Fall"

df["Season"] = df["Month"].apply(season)
```
