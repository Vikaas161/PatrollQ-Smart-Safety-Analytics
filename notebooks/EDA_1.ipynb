```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
crime_count = df["Primary Type"].value_counts().head(10)

plt.bar(crime_count.index, crime_count.values)
plt.title("Types of Crimes")
plt.xlabel("Crimes")
plt.ylabel("Count")
plt.xticks(rotation=70)
plt.show()
```

```python
df["Hour"].value_counts()
```

```python
hourly_crime = df.groupby("Hour").size()

plt.plot(hourly_crime, marker="o")
plt.xlabel("Hour")
plt.ylabel("Count")
plt.title("Crime by Hour")
plt.show()
```

```python
month_crime = df.groupby('Month').count()
plt.plot(month_crime, marker="o")
plt.xlabel("Month")
plt.ylabel("Count")
plt.title("Crime by Month")
plt.show()
```

```python
df["Arrest"].value_counts()
```

```python
arrest = df['Arrest'].value_counts()
plt.bar(['No Arrest', 'Arrest'], arrest.values)

plt.title("Arrest Distribution")
plt.ylabel("Number of Cases")
plt.show()
```

```python
weekend_crimes = df["Is_Weekend"].value_counts()

plt.bar(["Weekday", "Weekend"], weekend_crimes.values)
plt.title("Weekend vs Weekday Crime")
plt.ylabel("Number of Crimes")
plt.show()
```
